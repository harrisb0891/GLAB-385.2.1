## LIST EXRCISES

```python
# Create a function. take in 2 lists of years, reverse chronological order, add them together
# Create reusable codeblocks, can be reused as many times as needed
# CAN take input, AND can provide output
def reorder_profit_years(years1, years2):
    years1.reverse()
    years2.reverse()

    years1.extend(years2)

    return years1
```

```python
year_list1 = [2026,2025,2024, 2023]
year_list2 = [1990,1989,1988,1987]

result = reorder_profit_years(year_list1, year_list2)
print(result)
```
